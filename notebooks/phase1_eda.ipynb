### 3. Unified Representation & Token Mapping Demonstration

```python
from src.data.unified_schema import UnifiedRecord
from src.preprocessing.serialize import Serializer

# Load a sample record from each unified dataset
serializer = Serializer()
for ds in ['totto', 'dart', 'webnlg']:
    path = f'../data/unified/{ds}.jsonl'
    if os.path.exists(path):
        with open(path, 'r', encoding='utf-8') as f:
            line = json.loads(f.readline())
            rec = UnifiedRecord.from_dict(line)
            mapping = serializer.serialize_with_mapping(rec)
            print(f"\n=== [{ds.upper()}] Record ID: {rec.id} ===")
            print(f"Target Text: {rec.target_text}")
            print(f"Serialized (first 200 chars): {mapping['serialized_text'][:200]}...")
            print(f"Total Character Spans Mapped: {len(mapping['char_spans'])}")
```
